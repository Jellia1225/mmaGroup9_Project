```python
import random
import pandas as pd
from datetime import date, timedelta
```

```python
# ============================================================
# 1. GENERATING FLIGHT DATA
# ============================================================

# Flight origins and destinations
locations = ["Toronto", "Vancouver", "Montreal", "Quebec City", "Calgary", "Edmonton", "Ottawa", "Winnipeg"]

# The dates for the flights to happen: One year of trips, 2027
start_date = date(2027, 1, 1)
end_date = date(2027, 12, 31)
days_between = (end_date - start_date).days

flights = []

for i in range(1, 201):

    # Flight ID
    flight_id = f"PA{100 + i}"

    # Flight origins and destinations
    origin = random.choice(locations)
    destination = random.choice(locations)
    while destination == origin:
        destination = random.choice(locations)

    # Departure dates
    random_days = random.randint(0, days_between)
    departure_date = start_date + timedelta(days=random_days)

    # Season of the flight
    # This makes June, July, August, December and January vacation seasons
    # These vacation seasons will be used to determine if it will be more expensive to fly
    month = departure_date.month
    if month in [6, 7, 8, 12, 1]:
        season = "Vacation"
    else:
        season = "Regular"

    # Flight on weekend
    # This calculates if the day of the flight is a weekend or not
    is_weekend = departure_date.weekday() in [4, 5, 6]

    # Flight capacity
    # A reasonable capacity for airplanes in a domestic flight in Canada would be between 100 and 300 passengers
    capacity = random.randint(100, 300)

    # Seats remaining for the flight
    # The seats remaining for flights needs to be between 0 and the capacity of a given plane
    seats_remaining = random.randint(0, capacity)

    # Route demand
    # This is done so that flights with higher route demand are made more expensive
    route_demand = round(random.uniform(0, 1), 2)
    # This indicates the demand based on low, moderate, or high route demand, if needed
    #if route_demand < 0.33:
    #    demand_level = "Low"
    #elif route_demand < 0.66:
    #    demand_level = "Moderate"
    #else:
    #    demand_level = "High"

    # Flight base fares
    base_fare = round(random.uniform(200, 600), 2)

    # Dictionary for this flight
    flight_data = {
        "flight_id": flight_id,
        "origin": origin,
        "destination": destination,
        "departure_date": departure_date,
        "base_fare": base_fare,
        "seats_remaining": seats_remaining,
        "capacity": capacity,
        "route_demand": route_demand,
        "season": season,
        "is_weekend": is_weekend
    }   

    # Add flight to our dataset
    flights.append(flight_data)



# ============================================================
# 2. CREATING FLIGHT CLASS
# ============================================================

class Flight:

    def __init__(
        self,
        flight_id,
        origin,
        destination,
        departure_date,
        base_fare,
        seats_remaining,
        capacity,
        route_demand,
        season,
        is_weekend
    ):
        self.flight_id = flight_id
        self.origin = origin
        self.destination = destination
        self.departure_date = departure_date
        self.base_fare = base_fare
        self.seats_remaining = seats_remaining
        self.capacity = capacity
        self.route_demand = route_demand
        self.season = season
        self.is_weekend = is_weekend

# ============================================================
# 3. CONVERT DICTIONARIES INTO FLIGHT OBJECTS
# ============================================================

flight_objects = []
for flight_data in flights:

    flight = Flight(**flight_data)

    flight_objects.append(flight)

```

```python
# ============================================================
# 4. TEST
# ============================================================

print("Number of flight dictionaries:", len(flights))

print("Number of Flight objects:", len(flight_objects))

print()

print("First flight:")

print("Flight ID:", flight_objects[0].flight_id)
print("Origin:", flight_objects[0].origin)
print("Destination:", flight_objects[0].destination)
print("Departure date:", flight_objects[0].departure_date)
print("Base fare:", flight_objects[0].base_fare)
print("Seats remaining:", flight_objects[0].seats_remaining)
print("Capacity:", flight_objects[0].capacity)
print("Route demand:", flight_objects[0].route_demand)
print("Season:", flight_objects[0].season)
print("Is weekend:", flight_objects[0].is_weekend)
```
